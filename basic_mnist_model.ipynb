# **Handwritten digit recognition model**

**a.)** Importing all the libraries

```python
import tensorflow as tf
from tensorflow import keras
import matplotlib.pyplot as plt
%matplotlib inline
import numpy as np
```

**b.)** Importing the dataset MNIST

```python
(x_train,y_train),(x_test,y_test) = keras.datasets.mnist.load_data()
```

```python
display(x_train[0])
len(x_train)
```

```python
display(y_train[:5])
x_train.shape
```

**c.)** Reshaping the data

```python
x_train_flatten = x_train.reshape(len(x_train), 28*28)
x_train_flatten.shape
```

```python
#x_test.shape
x_test_flatten = x_test.reshape(len(x_test),28*28)
x_test_flatten.shape
```

```python
x_train_flatten = x_train_flatten / 255.0
x_test_flatten = x_test_flatten / 255.0

print("Normalized x_train_flatten (first 5 values):", x_train_flatten[0][:5])
```

**d.)** Model building

```python
model = tf.keras.models.Sequential([
  tf.keras.layers.Dense(10,input_shape=(784,), activation='relu'),
  tf.keras.layers.Dropout(0.2),
  tf.keras.layers.Dense(10)
])
```

```python
model = tf.keras.models.Sequential([
  tf.keras.layers.Dense(128, input_shape=(784,), activation='relu'),
  tf.keras.layers.Dropout(0.2),
  tf.keras.layers.Dense(64, activation='relu'), # Added another hidden layer
  tf.keras.layers.Dropout(0.2), # Added dropout for the new layer
  tf.keras.layers.Dense(10) # Output layer, no activation here because from_logits=True in loss_fn
])
```

```python
loss_fn = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True)
```

```python
model.compile(optimizer='adam',
              loss=loss_fn,
              metrics=['accuracy'])
```

**e.)** Model training

```python
model.fit(x_train_flatten, y_train, epochs=25)
```

```python
model.evaluate(x_test_flatten, y_test, verbose=2)
```

```python
model.save('mnist_model.h5')
```
